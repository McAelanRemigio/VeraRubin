### In this notebook we will query Gaia DR3's pulsating variable star catalogs specifically for the 7 regions covered by Rubin's DP1
### We will then match these labeled stars to DP1 (by their coordinates) so that we can assign labels to objects with more extensive data capture within DP1

```python
#Imports
from astroquery.gaia import Gaia
from astroquery.vizier import Vizier
from astropy.coordinates import SkyCoord
import astropy.units as u
import csv
import pandas as pd
from lsst.rsp import get_tap_service

from sklearn import svm
```

```python
#get an instance of the TAP service and assert that it exists

service = get_tap_service("tap")
assert service is not None
```

**Creating a function so that we can query each region/type with one line**

**We originally attempted to do this with 3 queries (one for each pulsating variable star class), however, we were unable to properly assign region labels through pandas and so this is an approach that we're confident in**

**That being said, it is fairly inefficient so it might be better to look into an approach that uses less queries**

```python
def queryGaia(variable_star_type, region, radius):
    if variable_star_type == 'RRLyrae':
        source = 'gaiadr3.vari_rrlyrae'
    elif variable_star_type == 'Cepheid':
        source = 'gaiadr3.vari_cepheid'
    elif variable_star_type == 'LPV':
       source = 'gaiadr3.vari_long_period_variable'

    # 47 Tuc Globular Cluster | 00h24m05s -72d04m48s | 6.02, -72.08
    # Low Ecliptic Latitude Field | 02h31m26s +06d58m48s | 37.86, 6.98
    # Fornax Dwarf Spheroidal Galaxy | 02h40m00s -34d27m00s | 40.00, -34.4
    # Extended Chandra Deep Field South (ECDFS) | 03h32m31s -28d06m00s | 53.13, -28.10
    # Euclid Deep Field South (EDFS) | 03h56m24s -48d43m48s | 59.10, -48.73
    # Low Galactic Latitude Field | 06h20m00s -25d00m00s | 95.00, -25.00
    # Seagull Nebula | 07h04m55s -10d30m36s | 106.23, -10.51
    if region == '47':
        center_ra = '6.02'
        center_dec = '-72.08'
    elif region == 'LELF':
        center_ra = '37.86'
        center_dec = '6.98'
    elif region == 'Fornax':
        center_ra = '40.00'
        center_dec = '-34.4'
    elif region == 'ECDFS':
        center_ra = '53.13'
        center_dec = '-28.10'
    elif region == 'EDFS':
        center_ra = '59.10'
        center_dec = '-48.73'
    elif region == 'LGLF':
        center_ra = '95.00'
        center_dec = '-25.00'
    elif region == 'Seagull':
        center_ra = '106.23'
        center_dec = '-10.51'

    # print('center_ra: ' + str(center_ra))
    # print('center_dec: ' + str(center_dec))
    
    query = f'''SELECT TOP 1000 s.source_id, g.ra, g.dec
            FROM {source} AS s
            JOIN gaiadr3.gaia_source AS g
            ON s.source_id = g.source_id
            WHERE CONTAINS(POINT('ICRS', ra, dec), CIRCLE('ICRS', {center_ra}, {center_dec}, {radius})) = 1
            '''

    job_gaia = Gaia.launch_job_async(query)
    df_gaia = job_gaia.get_results().to_pandas()
    return df_gaia
```

**Running the queries**

```python
df_47_rrl = queryGaia('RRLyrae', '47', 1)
df_47_ceph = queryGaia('Cepheid', '47', 1)
df_47_lpv = queryGaia('LPV', '47', 1)

df_LELF_rrl = queryGaia('RRLyrae', 'LELF', 1)
df_LELF_ceph = queryGaia('Cepheid', 'LELF', 1)
df_LELF_lpv = queryGaia('LPV', 'LELF', 1)

df_Fornax_rrl = queryGaia('RRLyrae', 'Fornax', 1)
df_Fornax_ceph = queryGaia('Cepheid', 'Fornax', 1)
df_Fornax_lpv = queryGaia('LPV', 'Fornax', 1)

df_ECDFS_rrl = queryGaia('RRLyrae', 'ECDFS', 1)
df_ECDFS_ceph = queryGaia('Cepheid', 'ECDFS', 1)
df_ECDFS_lpv = queryGaia('LPV', 'ECDFS', 1)

df_EDFS_rrl = queryGaia('RRLyrae', 'EDFS', 1)
df_EDFS_ceph = queryGaia('Cepheid', 'EDFS', 1)
df_EDFS_lpv = queryGaia('LPV', 'EDFS', 1)

df_LGLF_rrl = queryGaia('RRLyrae', 'LGLF', 1)
df_LGLF_ceph = queryGaia('Cepheid', 'LGLF', 1)
df_LGLF_lpv = queryGaia('LPV', 'LGLF', 1)

df_Seagull_rrl = queryGaia('RRLyrae', 'Seagull', 1)
df_Seagull_ceph = queryGaia('Cepheid', 'Seagull', 1)
df_Seagull_lpv = queryGaia('LPV', 'Seagull', 1)

print('All queries FINISHED')
```

**Assigning region labels so that we can identify where each star came from later on**

```python
df_47_rrl['Region'] = '47'
df_47_ceph['Region'] = '47'
df_47_lpv['Region'] = '47'

df_LELF_rrl['Region'] = 'LELF'
df_LELF_ceph['Region'] = 'LELF'
df_LELF_ceph['Region'] = 'LELF'

df_Fornax_rrl['Region'] = 'Fornax'
df_Fornax_ceph['Region'] = 'Fornax'
df_Fornax_lpv['Region'] = 'Fornax'

df_ECDFS_rrl['Region'] = 'ECDFS'
df_ECDFS_ceph['Region'] = 'ECDFS'
df_ECDFS_lpv['Region'] = 'ECDFS'

df_EDFS_rrl['Region'] = 'EDFS'
df_EDFS_rrl['Region'] = 'EDFS'
df_EDFS_rrl['Region'] = 'EDFS'

df_LGLF_rrl['Region'] = 'LGLF'
df_LGLF_ceph['Region'] = 'LGLF'
df_LGLF_lpv['Region'] = 'LGLF'

df_Seagull_rrl['Region'] = 'Seagull'
df_Seagull_ceph['Region'] = 'Seagull'
df_Seagull_lpv['Region'] = 'Seagull'
```

**Combining all stars of the same class (rrl, ceph, lpv) into one dataframe for each class**

```python
df_gaia_rrl = pd.concat([df_47_rrl, df_LELF_rrl, df_Fornax_rrl, df_ECDFS_rrl, df_EDFS_rrl, df_LGLF_rrl, df_Seagull_rrl])
df_gaia_ceph = pd.concat([df_47_ceph, df_LELF_ceph, df_Fornax_ceph, df_ECDFS_ceph, df_EDFS_ceph, df_LGLF_ceph, df_Seagull_ceph])
df_gaia_lpv = pd.concat([df_47_lpv, df_LELF_lpv, df_Fornax_lpv, df_ECDFS_lpv, df_EDFS_lpv, df_LGLF_lpv, df_Seagull_lpv])
```

**Quering RSP diaObject so that we can match our labeled pulsating variable stars to diaObjects**

```python
#RSP Query
query = """SELECT diaObjectId, ra, dec, g_psfFluxMean, r_psfFluxMean, i_psfFluxMean,
        g_psfFluxSigma, r_psfFluxSigma, i_psfFluxSigma,
        g_psfFluxChi2, r_psfFluxChi2, i_psfFluxChi2,
        g_psfFluxStetsonJ, r_psfFluxStetsonJ, i_psfFluxStetsonJ,
        g_psfFluxMAD, r_psfFluxMAD, i_psfFluxMAD,
        g_psfFluxNdata, r_psfFluxNdata, i_psfFluxNdata, nDiaSources
        FROM dp1.DiaObject
        WHERE ra IS NOT NULL
        AND dec IS NOT NULL
        """

job = service.submit_job(query)
job.run()
job.wait(phases=['COMPLETED', 'ERROR'])
print('Job phase is', job.phase)
if job.phase == 'ERROR':
    job.raise_if_error()
assert job.phase == 'COMPLETED'
df_rsp = job.fetch_result().to_table().to_pandas() #get results -> turn into table -> turn table into pandas dataframe
print(str(len(df_rsp)))
```

**Matching coordinates**

Any object within 1 arcsecond we are confident enough in that we are considering it a "match"

```python
#Converting RA/DEC from RSP to SkyCoord Objects
coords_rsp = SkyCoord(ra=df_rsp['ra'].values*u.deg, dec=df_rsp['dec'].values *u.deg)
```

```python
#Converting RA/DEC from gaia rrl/ceph/lpv to SkyCoord objects
df_gaia_rrl['ra'] = pd.to_numeric(df_gaia_rrl['ra'], errors='coerce')
df_gaia_rrl['dec'] = pd.to_numeric(df_gaia_rrl['dec'], errors='coerce')

df_gaia_ceph['ra'] = pd.to_numeric(df_gaia_ceph['ra'], errors='coerce')
df_gaia_ceph['dec'] = pd.to_numeric(df_gaia_ceph['dec'], errors='coerce')

df_gaia_lpv['ra'] = pd.to_numeric(df_gaia_lpv['ra'], errors='coerce')
df_gaia_lpv['dec'] = pd.to_numeric(df_gaia_lpv['dec'], errors='coerce')

coords_gaia_rrl = SkyCoord(ra=df_gaia_rrl['ra'].values*u.deg, dec=df_gaia_rrl['dec'].values*u.deg)
coords_gaia_ceph = SkyCoord(ra=df_gaia_ceph['ra'].values*u.deg, dec=df_gaia_ceph['dec'].values*u.deg)
coords_gaia_lpv = SkyCoord(ra=df_gaia_lpv['ra'].values*u.deg, dec=df_gaia_lpv['dec'].values*u.deg)
```

```python
idx_rrl, d2d_rrl, _ = coords_gaia_rrl.match_to_catalog_sky(coords_rsp)
matches_rrl = d2d_rrl < 1* u.arcsec
matched_gaia_rrl = df_gaia_rrl[matches_rrl].reset_index(drop=True)
matched_diaObjects_rrl = df_rsp.iloc[idx_rrl[matches_rrl]].reset_index(drop=True)
matched_diaObjects_rrl['Region'] = matched_gaia_rrl["Region"]

idx_ceph, d2d_ceph, _ = coords_gaia_ceph.match_to_catalog_sky(coords_rsp)
matches_ceph = d2d_ceph < 1* u.arcsec
matched_gaia_ceph = df_gaia_ceph[matches_ceph].reset_index(drop=True)
matched_diaObjects_ceph = df_rsp.iloc[idx_ceph[matches_ceph]].reset_index(drop=True)
matched_diaObjects_ceph['Region'] = matched_gaia_ceph["Region"]

idx_lpv, d2d_lpv, _ = coords_gaia_lpv.match_to_catalog_sky(coords_rsp)
matches_lpv = d2d_lpv < 1* u.arcsec
matched_gaia_lpv = df_gaia_lpv[matches_lpv].reset_index(drop=True)
matched_diaObjects_lpv = df_rsp.iloc[idx_lpv[matches_lpv]].reset_index(drop=True)
matched_diaObjects_lpv['Region'] = matched_gaia_lpv["Region"]
```
