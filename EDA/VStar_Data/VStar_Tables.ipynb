### VStar Tables

We are now going to combine the all of the tables that we gathered of RRLyrae, Cephied, and LPV stars located in Fornax into a centralized table to utilize for validation of our methods and period/magnitude/distance analysis with light curves. We will clean and filter out bad values and non-relevant columns as well, and check for duplicates and overlapping detections in each list.


Columns of Interest From Each Table:
1. Sky Coordinates: right ascension (deg), declination(deg)
2. Periods (if known), NaN if not
3. E(B-V) --> reddening?
4. Flux Magnitudes (V/I bands, deviations)
5. Amplitudes
6. Classification of variable star (e.g. isCStar for LPV, p2c/anomalous cepheids, not as relevant for rrlyrae)

** For Gaia look into other columns

### Import necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import astroquery.gaia as Gaia
```

### Data Cleaning / Merging of Variable Star Tables

```python
# retrieve current directory
dir = os.getcwd()
print(dir)
```

```python
# retrieve each of tables and convert to pandas dataframes
# Appreviations: bw - Bersier Wood, b - Braga et. al.

bw_rrl_path = os.path.join(dir, 'VStar_Tables', 'RRLyrae_Bersier_Wood.csv')
bw_ceph_path = os.path.join(dir, 'VStar_Tables', 'Cepheid_Bersier_Wood.csv')
bw_lpv_path = os.path.join(dir, 'VStar_Tables', 'LPV_Cand_Bersier_Wood.csv')
b_rrl_path = os.path.join(dir, 'VStar_Tables', 'Braga_RRLyrae.csv')

bw_rrl_df = pd.read_csv(bw_rrl_path)
bw_ceph_df = pd.read_csv(bw_ceph_path)
bw_lpv_df = pd.read_csv(bw_lpv_path)
b_rrl_df = pd.read_csv(b_rrl_path)

# create an empty dataframe for the merged table
v_stars_df = pd.DataFrame()
```

```python
# look at what each of the tables look like to ensure translated properly
print('Bersier Wood - RR Lyrae\n' + str(bw_rrl_df.head()) + '\n')
print(bw_rrl_df.shape)
```

```python
print('Bersier Wood - Cepheid\n' + str(bw_ceph_df.head()) + '\n')
print(bw_ceph_df.shape)
```

```python
print('Bersier Wood - LPV\n' + str(bw_lpv_df.head()) + '\n')
print(bw_lpv_df.shape)
```

```python
print('Braga - RR Lyrae\n' + str(b_rrl_df.head()) + '\n')
print(b_rrl_df.shape)
```

```python
# the braga table has star data from a different galaxy which we don't want
# we only want fornax data
b_rrl_df = b_rrl_df[b_rrl_df['galaxy'] == 'F']

# check this worked
print(b_rrl_df['galaxy'].unique())
b_rrl_df.shape
```

```python
# extract the columns we want which include coordinates, period, flux magnitudes, their errors and any color measurements
# Performed through either choosing or dropping specific columns
bw_rrl_subset = bw_rrl_df[['_RA','_DE', 'Period', 'Vmag', 'e_Vmag', 'Icmag', 'e_Icmag', 'E(B-V)']]
bw_ceph_subset = bw_ceph_df[['_RA','_DE', 'Per', 'Class', 'Vmag', 'e_Vmag', 'V-Ic','e_V-Ic', 'E(B-V)']]
bw_lpv_subset = bw_lpv_df[['_RA', '_DE', 'Vmag', 'e_Vmag', 'V-Ic', 'e_V-Ic', 'E(B-V)']]
b_rrl_subset = b_rrl_df.drop(['galaxy', 'name', 'dominant_mode'], axis=1)
```

```python
# ensure the tables look as intended
print('Bersier-Wood RR Lyrae ' + str(bw_rrl_subset.shape))
print('Bersier-Wood Cepheid ' + str(bw_ceph_subset.shape))
print('Bersier-Wood LPV ' + str(bw_lpv_subset.shape))
print('Braga et. al. RRLyrae ' + str(b_rrl_subset.shape))
```

```python
# check for any rows with missing values
print(bw_rrl_subset.isnull().sum(), '\n')
print(bw_ceph_subset.isnull().sum(), '\n')
print(bw_lpv_subset.isnull().sum(), '\n')
print(b_rrl_subset.isnull().sum(), '\n') # has some missing values for flux
```

```python
# first rename the columns for consistency
bw_rrl_subset = bw_rrl_subset.rename(columns={'_RA': 'ra', '_DE': 'dec', 'Period': 'period'})
bw_ceph_subset = bw_ceph_subset.rename(columns={'_RA': 'ra', '_DE': 'dec', 'Per': 'period', 'Class': 'class'})
bw_lpv_subset = bw_lpv_subset.rename(columns={'_RA': 'ra', '_DE': 'dec'})
b_rrl_subset = b_rrl_subset.rename(columns={'ra': 'ra', 'dec': 'dec', 'per_fu': 'period'})

# create a dict for renaming columns for the braga table
renamed_columns = {'vmag': 'Vmag', 'errvmag': 'e_Vmag', 'errbmag': 'e_bmag', 'errrmag': 'e_rmag', 
                   'errimag': 'e_imag', 'per_fo': 'period_fo', 'per_fu': 'period_fu',
                   'ampv': 'vamp', 'err_ampv': 'e_vamp', 'ampb': 'bamp', 'err_bamp': 'e_bamp',
                   'ampr': 'ramp', 'err_ampr': 'e_ramp', 'ampi': 'iamp', 'err_ampi': 'e_iamp', 
                   'ampv_sec': 'vamp_sec', 'err_ampv_sec': 'e_vamp_sec', 'ampb_sec': 'bamp_sec', 
                   'err_ampb_sec': 'e_bamp_sec'
                }

b_rrl_subset = b_rrl_subset.rename(columns=renamed_columns)
```

```python
# add columns for the paper / source the data was recieved from
bw_rrl_subset['source'] = 'Bersier Wood'
bw_ceph_subset['source'] = 'Bersier Wood'
bw_lpv_subset['source'] = 'Bersier Wood'
b_rrl_subset['source'] = 'Braga'

# add column for 'main category' representing classification of variable star
bw_rrl_subset['main category'] = 'RRLyrae'
bw_ceph_subset['main category'] = 'Cepheid'
bw_lpv_subset['main category'] = 'LPV'
b_rrl_subset['main category'] = 'RRLyrae'
```

```python
v_stars_df = pd.concat([bw_rrl_subset, bw_ceph_subset, bw_lpv_subset, b_rrl_subset], ignore_index=True)
```

```python
v_stars_df
```

```python
# lets combine this table with the Gaia data
# get paths
gaia_lpv_path = os.path.join(dir, 'VStar_Tables', 'Gaia_lpv.csv')
gaia_rrl_path = os.path.join(dir, 'VStar_Tables', 'Gaia_rrl.csv')

# convert from csv to pd dataframe
gaia_lpv_df = pd.read_csv(gaia_lpv_path)
gaia_rrl_df = pd.read_csv(gaia_rrl_path)

print(gaia_lpv_df.head(), '\n')
print(gaia_rrl_df.head())
```

```python
gaia_rrl_df = gaia_rrl_df.rename(columns={'pf': 'period', 'int_average_bp': 'bpmag', 'int_average_g':'gmag', 
                                          'int_average_rp': 'rpmag', 'bp_rp':'bp-rp'})

# create a new column in gaia lpv for period, calculated by taking inverse of frequency
gaia_lpv_df['period'] = 1 / gaia_lpv_df['frequency']
gaia_lpv_df['class'] = gaia_lpv_df['is_cstar'].apply(lambda x: 'CStar' if x == True else 'O-rich Star')

gaia_lpv_df = gaia_lpv_df.rename(columns={'bp_rp': 'bp-rp', 'phot_g_mean_mag': 'gmag', 'phot_bp_mean_mag': 'bpmag', 'phot_rp_mean_mag': 'rpmag'})
```

```python
gaia_lpv_df['period']
```

```python
gaia_lpv_df['class']
```

```python
# get subsets of interest, including coordinates, period, flux magnitudes, and color
gaia_rrl_subset = gaia_rrl_df[['ra', 'dec', 'period', 'gmag', 'bpmag', 'rpmag', 'bp-rp']]
gaia_lpv_subset = gaia_lpv_df[['ra', 'dec', 'period', 'class', 'gmag', 'bpmag', 'rpmag', 'bp-rp']]
```

```python
# create columns for source and main category of star
gaia_rrl_subset['source'] = 'Gaia'
gaia_lpv_subset['source'] = 'Gaia'

gaia_rrl_subset['main category'] = 'RRLyrae'
gaia_lpv_subset['main category'] = 'LPV'
```

```python
print(gaia_lpv_subset.head())
print(gaia_lpv_subset.shape)
```

```python
print(gaia_rrl_subset.head())
print(gaia_rrl_subset.shape)
```

```python
# see null values for each
print(gaia_rrl_subset.isnull().sum(), '\n')
print(gaia_lpv_subset.isnull().sum(), '\n') # can see many lpvs have missing periods which is expected
```

```python
v_stars_df = pd.concat([v_stars_df, gaia_rrl_subset, gaia_lpv_subset], ignore_index=True)
```
